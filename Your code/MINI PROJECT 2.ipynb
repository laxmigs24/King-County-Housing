```python
import seaborn as sns
import matplotlib.pyplot as plt
import numpy as np

# 1. Calculate the correlation matrix
corr_matrix = data.corr(numeric_only=True)

# 2. Plotting the Heatmap
plt.figure(figsize=(10, 8))
mask = np.triu(np.ones_like(corr_matrix, dtype=bool)) # Mask the upper triangle for clarity
sns.heatmap(corr_matrix, mask=mask, annot=True, fmt=".2f", cmap='coolwarm', linewidths=0.5)
plt.title('Correlation Heatmap of King County Housing Data')
plt.show()

# 3. Correlation with the Target (Price) specifically
price_correlations = corr_matrix['price'].sort_values(ascending=False)
print("Correlation of Features with Price:\n", price_correlations)
```

Analysis

1. Features like sqft_living, grade, and sqft_above usually show the highest correlation with price (often $> 0.70$).

2. This tells us that the physical size and the quality of construction materials are the primary drivers of value in King County.

3. The correlation between sqft_living and sqft_above or bathrooms. There is a very high correlation (e.g., $0.75$ to $0.85$).

4. Features like zipcode, condition, or yr_built often show surprisingly low linear correlation with price.
